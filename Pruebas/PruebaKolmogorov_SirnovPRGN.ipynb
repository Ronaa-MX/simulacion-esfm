```python
#Librerías a utilizar
import numpy as np
from scipy.stats import kstest
```

```python
#Función para realizar el test contra la distribución uniforme
def ks_test(valores):
  """
    Realiza una prueba de Kolmogorov-Smirnov  para determinar si los datos
    provienen de una distribución uniforme en [0, 1).

    Parámetros:
    -----------
    valores : list o np.array
        Lista de valores numéricos en el rango [0, 1).
  """
  ks_statistic, p_value = kstest(valores, 'uniform', args=(0, 1))

  print(f"Estadístico KS: {ks_statistic:.4f}")
  print(f"Valor p: {p_value:.4f}")

  # Interpretación
  alpha = 0.05
  if p_value < alpha:
      print("Rechazar la hipótesis nula: los datos no siguen la distribución uniforme.")
  else:
      print("No rechazar la hipótesis nula: los datos siguen la distribución uniforme.")

ks_test(lista_prn)
```
