## Prueba de Rachas
Realizaremos una prueba de Rachas con los números generados aleatoriamente. Con el fin de revisar si estos son independientes entre sí, siendo que tendremos que:
 $$H_0:\text{Los númerosde la muestra } \{x_{i}\} \text{ son independientes}$$
 $$H_a:\text{Los númerosde la muestra } \{x_{i}\} \text{ no son independientes}$$

```python
import math
from scipy.stats import norm

def test_rachas(sequence, alpha=0.05):
    """
    Prueba de rachas de subidas y bajadas para una secuencia numérica.

    Parámetros:
    sequence : list of float
        Lista de números a evaluar.
    alpha : float, opcional
        Nivel de significancia para la interpretación (default 0.05).

    Retorna:
    dict: Resultados detallados con estadísticos, p-valor e interpretación.
    """
    # Validaciones
    if not isinstance(sequence, list):
        raise TypeError("La secuencia debe ser una lista.")
    n = len(sequence)
    if n < 2:
        raise ValueError("Se requieren al menos 2 elementos.")

    # Construir secuencia de signos (1=subida, 0=bajada)
    signs = []
    for i in range(n - 1):
        diff = sequence[i+1] - sequence[i]
        if diff > 0:
            signs.append(1)
        elif diff < 0:
            signs.append(0)
        else:
            raise ValueError("Valores iguales consecutivos no permitidos.")

    # Contar rachas (método óptimo, una pasada)
    m = len(signs)
    if m == 0:
        raise ValueError("No hay suficientes datos después de eliminar empates.")

    R = 1
    for i in range(1, m):
        if signs[i] != signs[i-1]:
            R += 1

    # Parámetros teóricos (usando n = número original de observaciones)
    media = (2 * n - 1) / 3.0
    varianza = (16 * n - 29) / 90.0
    sigma = math.sqrt(varianza)

    # Estadístico Z y p-valor
    Z = (R - media) / sigma
    p_valor = 2 * (1 - norm.cdf(abs(Z)))

    # Valor crítico para alpha bilateral
    z_critico = norm.ppf(1 - alpha/2)

    # Interpretación
    if abs(Z) < z_critico:
        decision = "No se rechaza H0"
        conclusion = "La secuencia es compatible con la aleatoriedad (no hay evidencia de patrón)."
    else:
        decision = "Se rechaza H0"
        conclusion = "La secuencia no es aleatoria (presenta tendencia o patrón significativo)."

    # Advertencia por tamaño de muestra
    if n < 20:
        warning = "Advertencia: n es pequeño (<20). La aproximación normal puede no ser precisa."
    else:
        warning = None

    # Resultados formateados
    resultados = {
        'n': n,
        'R': R,
        'media_teorica': round(media, 4),
        'sigma': round(sigma, 4),
        'Z': round(Z, 4),
        'p_valor': round(p_valor, 4),
        'alpha': alpha,
        'z_critico': round(z_critico, 4),
        'decision': decision,
        'conclusion': conclusion,
        'warning': warning
    }
    return resultados
```
