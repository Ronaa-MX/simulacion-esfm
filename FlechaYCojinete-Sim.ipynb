## Problema

Se tiene una flecha que será ensamblada en un cojinete como se muestra en la figura.

Si **$x_1$** sigue una distribución normal con media $1.5$ y variancia $0.0016$ y si **$x_2$** sigue esta misma distibución con media $1.48$ y varianza $0.0009$, determinar:

*  La probabilidad de que haya interferencia
* El nùmero de veces necesario que es necesario simular el experimento, si se quiere que la probabilidad de interfencia estimada difiera de su valor verdadero en menos de $0.01$, con un nivel de seguridad de $95\%$

### Simulaciòn de la probabilidad de interfencia

```python
## librerías
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

```python
#Distribución normal por 
def u_suma():
    total = 0
    for i in range(12):
        suma = r.random()
        total += suma
    return total

def gauss(mu,dest,n):
    z_val =[]
    total=0
    for i in range(n):
        x = mu+dest*(u_suma()-6)
        z=(x-mu)/dest
        z_val.append(z)
    return z_val
```

```python
#Experimento
def inteference(mu1,dest1,mu2,dest2,n):
    x = gauss(mu1,dest1,n)
    y = gauss(mu2,dest2,n)
    interferencia = []
    for i in range(n):
        if x[i]<y[i]:
            interferencia.append(1)
        else:
            interferencia.append(0)
    return np.sum(interferencia)/np.size(interferencia)
```
