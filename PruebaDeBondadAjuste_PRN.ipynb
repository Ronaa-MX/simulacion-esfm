## Prueba de Bondad de Ajuste

Realizaremos una prueba de bondad de ajuste, con los números generados aleatoriamente. Con el fin de revisar si estos son de una distribución uniforme, siendo que tendremos que $$H_0: x_{n}\sim U(0,1)$$ $$H_a:x_{n}\nsim U(0,1)$$

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency, uniform

# Valores generados para una distribución Uniforme
prn = [0.78961, 0.05230, 0.10699, 0.55877, 0.14151,
    0.76086, 0.12079, 0.27738, 0.65726, 0.79269,
    0.80548, 0.82654, 0.29453, 0.20852, 0.42989,
    0.58518, 0.98611, 0.34488, 0.34358, 0.11537,
    0.89898, 0.57880, 0.67621, 0.05010, 0.00121,
    0.28269, 0.73059, 0.70119, 0.18284, 0.49962,
    0.38618, 0.76910, 0.68334, 0.55170, 0.10850,
    0.79882, 0.45679, 0.21631, 0.87616, 0.55743,
    0.58962, 0.33216, 0.03185, 0.61168, 0.09264,
    0.69623, 0.17028, 0.05475, 0.91512, 0.76262,
    0.29931, 0.30861, 0.83358, 0.51781, 0.03272,
    0.57410, 0.26593, 0.85903, 0.43308, 0.35286,
    0.24000, 0.65559, 0.38507, 0.90829, 0.94187,
    0.93655, 0.88809, 0.81772, 0.36982, 0.19904,
    0.54325, 0.62400, 0.09133, 0.41678, 0.33954,
    0.58244, 0.85853, 0.86752, 0.38729, 0.15506,
    0.23949, 0.53559, 0.33381, 0.49883, 0.75103,
    0.19962, 0.65002, 0.74579, 0.79113, 0.63453,
    0.19147, 0.40644, 0.08128, 0.73435, 0.22724,
    0.22287, 0.07281, 0.64183, 0.44267, 0.72102]
```

```python
def prueba_bondad_ajuste_uniforme(datos, n_intervalos=10, alpha=0.05):
    """
    Realiza una prueba de bondad de ajuste de Chi-cuadrado para determinar si los datos
    provienen de una distribución uniforme en [0, 1).

    Parámetros:
    -----------
    datos : list o np.array
        Lista de valores numéricos en el rango [0, 1).
    n_intervalos : int, opcional
        Número de intervalos (bins) para dividir los datos (por defecto: 10).
    alpha : float, opcional
        Nivel de significancia para la prueba (por defecto: 0.05).
    """

    # Convertir los datos a un array de numpy
    datos = np.array(datos)

    # Definir los límites de los intervalos (bins)
    bin_edges = np.linspace(0, 1, n_intervalos + 1)

    # Calcular las frecuencias observadas
    frecuencias_observadas, _ = np.histogram(datos, bins=bin_edges)

    # Calcular las frecuencias esperadas (uniforme: todas iguales)
    n_datos = len(datos)
    frecuencias_esperadas = np.full(n_intervalos, n_datos / n_intervalos)

    # Realizar la prueba de Chi-cuadrado
    chi2, p_valor, _, _ = chi2_contingency([frecuencias_observadas])

    # Comparar el valor p con el nivel de significancia (alpha)
    rechazar_h0 = p_valor < alpha

    # Crear la tabla de frecuencias
    tabla_frecuencias = pd.DataFrame({
        'Intervalo': [f"[{bin_edges[i]:.2f}, {bin_edges[i+1]:.2f})" for i in range(n_intervalos)],
        'Frecuencia Observada': frecuencias_observadas,
        'Frecuencia Esperada': frecuencias_esperadas
    })

    # Resultados
    resultados = {
        'Estadístico Chi-cuadrado': chi2,
        'Valor p': p_valor,
        '¿Rechazar H0?': rechazar_h0,
        'Conclusión': "Los datos NO siguen una distribución uniforme." if rechazar_h0 else "Los datos SÍ siguen una distribución uniforme."
    }

    return resultados, tabla_frecuencias

resultados, tabla = prueba_bondad_ajuste_uniforme(prn, n_intervalos=5, alpha=0.05)

print("Resultados de la prueba:")
for clave, valor in resultados.items():
    print(f"{clave}: {valor}")

print("\nTabla de frecuencias:")
print(tabla)
```
