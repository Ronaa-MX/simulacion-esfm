# Generador Congruencial

```python
import tabulate
def generador(a,b,m,x_0):
  x_i= (a*x_0+b)%m
  return x_i
```

```python
n= 70
a= 5
b= 5
m= 64
x_0=1

muestra=[]
datos=[]

for i in range(n):
 nuevo = generador(a,b,m,x_0)
 muestra.append([i,x_0,nuevo/64])
 datos.append(x_0/64)
 x_0=nuevo

encabezados=["i","x_i","X_{i+1}"]
print(tabulate.tabulate(muestra, headers=encabezados, tablefmt="psql"))
```
