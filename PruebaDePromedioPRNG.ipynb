```python
#Librerías a importar para realizar la prueba de hipótesis
import statistics as st
import numpy as np
import math
import matplotlib.pyplot as plot
from scipy import stats
```

```python
#Función de prueba de hipótesis
def prueba_uniforme(valores,mu,std):
  media_barra = st.mean(valores)
  prn = np.array(valores)
  z_score = (media_barra - mu) / (math.sqrt(std)) / np.sqrt(len(valores))
  # Calcular valor p (prueba bilateral)
  p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

  print(f"Estadístico z: {abs(z_score):.4f}")
  print(f"Valor p: {p_value:.4f}")

  if p_value < 0.05:
    print("Rechazar la hipótesis nula.")
  else:
    print("No rechazar la hipótesis nula.")
  return media_barra
```

```python
#Prueba con la lista de valores dada
x_barra=prueba_uniforme(lista_prn,1/2,1/12)
```

```python
#Gráfica de valores dados,su media y la media de una distribución uniforme de 0 a 1
intervalo = np.linspace(0, 1, len(lista_prn))
fig, ax = plot.subplots()
ax.plot(intervalo,lista_prn, label='Números Pseudo-Aleatorios')
ax.axhline(y=x_barra, color='red', linestyle='--', label='Media de los Números Pseudo-Aleatorios')
ax.axhline(y=.5, color='blue', linestyle=':', label='Media real')
ax.legend()
plot.show()
```
