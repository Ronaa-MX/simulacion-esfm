## Generación de números pseudo-aleatorios de una distribución de Weibull

Generaremos números pseudo-aleatorios de una distribución de Weibull a través del método de la transformada de la función inversa. Partimos de que tenemos una distribución uniforme, donde sea $R\sim U(0,1)$. Entonces partiremos de suponer que la función de distribución acumulada sera igual a la de la función de distribución acumulada de la distibución objetivo, en este la distribución de Weibull con paramétros $\alpha$ y $\beta$, siendo $\beta=1$.
Siendo la distribución de Weibull con los paramétros dados:
$$
We(\alpha,1)=\begin{cases}
\alpha x^{\alpha-1}e^{-x^\alpha}&\text{, sí } x\geq 0 \\
0 \text{, otro caso}
\end{cases}
$$
De ahí tendremos que la función de distribución acumulada de la distribución de Weibull es $F(x)=1-e^{-x^{\alpha}}$, igualando y despejando $x$, se tiene que $x=\sqrt[\alpha]{\ln(\frac{1}{1-R})}$, que igualmente podría ser $x=\sqrt[\alpha]{\ln(\frac{1}{R})}$, dado que la distribución uniforme va de 0 a uno.

```python
## Librerías a utilizar
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

```python
# Función de generación de número pseudo-aleatorios de una distribución Weibull
def weibull_random(alpha,n):
    we = []
    for i in range(n):
        u = r.random()
        x = np.power(np.log(1/(1-u)), 1/alpha).item()
        we.append(x)
    return we
```

```python
#Histograma de la distribución de Weibull con los parametros dados
#Parametros
alpha = 2
n = 10000
plt.hist(weibull_random(alpha,n),bins=50)
plt.show()
```

```python
# Función de distribución acumulativa con los números aleatorios de la distribución de Weibull
weibull_random=np.sort(weibull_random(alpha,n))
weibull_standard = [x/n for x in weibull_random ]
acum_weibull = np.cumsum(weibull_standard,axis=None)
plt.plot(weibull_random,acum_weibull)
plt.title('$F(x)$ de Weibull')
plt.show()
```
